#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:

- Transform the data into a tidy form (if needed)

```r
tidyball <- WMM |>
    filter(!is.na(conf_wins_pct), !is.na(seed), !is.na(tourney_finish), seed !=0) |> 
    #removed the NA rows by reversing the logic of is.na. also removed the zero seed since it was a special occasion and would mess up KNN along with the na values

    mutate(top_16 = tourney_finish %in% #checks if each team finished in (at worst) the top 16, and returns true if Yes, False if no. then store these values in a new column called top_16
           c("top_16_loss", "top_8_loss", "top_4_loss", "top_2_loss", "champ")) |>  
    select(conf_wins_pct, seed, top_16) |> #select conference win percentages, seeds, top_16 and removed the irrelavent columns
    mutate(top_16 = as_factor(top_16)) #converted top_16 into a categorical variable instead of a logical variable 

tidyball
```

Visualization 1: Mean conference winning percentage against top16 advancement

```r
options(repr.plot.width = 12, repr.plot.height = 6)
confpercentages_summary<- tidyball|>
    group_by(top_16) |>
    summarize(mean_conf_wins = mean(conf_wins_pct))
confpercentages_summary

plot3 <- confpercentages_summary|>
    ggplot(aes(x=top_16, y = mean_conf_wins))+
    geom_bar(stat= "identity")+
    labs(x="Reached Top 16", y= "Mean Conference Winning Percentage", title = "Mean Conference Winning Percentage by top 16 advancement") +
    theme(text=element_text(size=14))
plot3
```

The visualization above compares the mean conference winning percentages of teams that advanced to the top 16 and those that did not. Teams that reached the top 16 had a higher mean conference winning percentage of about 82.156% compared to teams that were eliminated earlier at 74.982%. This menas that the conference percentage may be a useful predictor for the advancement to the top 16, since teams with a higher conference winning percentage may have better regular-season performance (which could mean better tournament rankings)

Visualization 2: Tournament seed against top16 advancement

```r
options(repr.plot.width = 12, repr.plot.height = 6)
seed_summary<- tidyball|>
    group_by(seed) |>
    summarize(proportion=mean(top_16=="TRUE"))
seed_summary

plot4 <- seed_summary|>
    ggplot(aes(x=as_factor(seed), y = proportion))+ #used chat to identify why seed did not work
    geom_bar(stat= "identity")+
    labs(x="Tournament Seed", y= "Proportion Reaching Top 16", title = "Top 16 Advanacement Proportion by Tournament Seed") +
    theme(text=element_text(size=14))
plot4
```

The visualization above shows the proportion of teams advancing to the top 16 based on their tournament seed. We see that smaller seed values (1,2,3 and 4) have significantly higher advancement proportions (about 95% of seed 1s made it to top 16), compared to a sudden drop in advancement (~29% for teams in seed 5 or above). Although there are some exceptions like near 11 and 12, the overal trend shows that tournament seed is a useful predictor for top16 advancement (since lower number seeds represent higher-ranked teams that are already expected to do good in the tournament). Also, the visualization shows a better representation compared to mean conference win percentages.
